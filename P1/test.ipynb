```python
import pydicom
from pathlib import Path
```

```python
dicom_file = pydicom.dcmread(r"LungCT-Diagnosis/1-01.dcm")
dicom_file
```

```python
pixel_arr = dicom_file.pixel_array
```

## Show Image 1

```python
import numpy as np
print(np.unique(pixel_arr))
```

```python
pixel_arr_float = pixel_arr.astype(float)
pixel_arr_float
```

```python
p_arr = np.maximum(pixel_arr_float, 0)
n_arr = np.divide(p_arr, p_arr.max())
n_arr = np.multiply(n_arr, 255.0)
n_arr = np.uint8(n_arr)
```

```python
import PIL
img = PIL.Image.fromarray(n_arr)
```

```python
img.show()
```
