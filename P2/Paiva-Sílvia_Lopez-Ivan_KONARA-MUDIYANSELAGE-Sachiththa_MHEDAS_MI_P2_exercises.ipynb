##2. Artifact correction and clean-up



*   Correcting artifacts and cleaning the image is crucial for the success of **automated processing methods** (i.e., segmentation, registration, ...)
*   A typical issue with MRI scans is caused by **spatial inhomogeneties** of the magnetic field, namely **bias field**, a smooth signal intensity variation within tissue of the same physical properties.
*   Note that **bias field** is modeled as a multiplicative factor, whereas ***noise*** is assumed to be an additive and spatially independent (Gaussian) factor.

###Bias field correction
There are many automated pipelines that can be used to perform this corrections (some are integrated within longer pipelines such as FreeSurfer, FSL and SPM).

In this example, we will *simulate* the use of the ANTs software. It can be run independently (see [documentation](https://http://stnava.github.io/ANTs/)) or within the nipype package.

- Setting `0` to `250` ignores a few extreme bright dots so the full color range stretches across normal brain tissue, making subtle shading differences easy to see.

**Exercise:** In your owns words, explain what is bias and how it affects images.

**Answer**: In an ideal MRI, the signal strength for a specific type of tissue should look the same everywhere in the image at a given time. In reality, the signal for that same tissue changes because of noise and bias. Bias is a **multiplying** effect on the signal that causes a smooth, gradual shift in brightness across the image, making the exact same tissue look brighter in some areas and darker in others.

## 4. Brain Segmentation
A common preprocessing step that separates the main brain tissues using the histogram.
There are many available tools to do this, for example FSL, FreeSurfer or SPM.

Here, we will explore the outputs of a basic tissue segmentation.

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
````

````python
# use imshow to plot the figures
# Plot first figure
T1_img = base_dir / 'example_T1w.nii.gz'
T1_img = nib.load(T1_img)
T1_data = T1_img.get_fdata()
plt.imshow(T1_data[:, :, 160], vmax=500, origin="lower")
plt.title('T1 data before skull-stripping')
````

Now, we can compute the histogram of this image:

````python

# Compute the optimal bin size according to the data
bins = np.histogram_bin_edges(T1_data[T1_data!= 0], bins='auto')
# Plot the histogram
sns.histplot(T1_data[T1_data != 0], bins=bins)
plt.xlim([0, 255])
````

**Question:** What do you observe from the previous histogram?

**Answer**: The histogram shows the non-zero brightness values of the flattened 3D MRI scan `T1_img` (combining the sagittal, axial, and coronal planes). There is a huge spike between `0` and `20` (reaching over `700,000`), which comes from background noise in the empty air around the head. Because this spike is so tall, it squashes the rest of the chart. As a result, the real brain tissue values (spread from `20` to `240`) look flat along the bottom, making it hard to see the different types of brain tissue clearly.

Now, we will examine the outputs of the segmentation together with the  histogram. Note that we have computed the segmentation before and we give the output to you:

````python
# NOTE: brain data is stored into 'T1_brain_data'
import matplotlib.pyplot as plt
import itertools


# Map segmentation label numbers to tissue names
tissue_names = {1: 'CSF', 2: 'Gray Matter', 3: 'White Matter'}

T1_brain_img = base_dir / 'example_T1w.nii.gz'
T1_brain_img = nib.load(T1_brain_img)
T1_brain_data = T1_brain_img.get_fdata()
T1_seg_image = base_dir / 'example_T1_brain_seg.nii.gz'
T1_seg_image = nib.load(T1_seg_image)
T1_seg_data = T1_seg_image.get_fdata()
labels = np.unique(T1_seg_data[T1_seg_data != 0]).astype('int')
palette = itertools.cycle(sns.color_palette("Set3"))
for label in labels:
    sns.histplot(T1_brain_data[T1_seg_data == label], bins=bins, color=next(palette), alpha=0.6, label=tissue_names.get(label, f'Label {label}'))
plt.legend(title='Tissue Type')
plt.show()
````

**Exercise**: Try different options of the plotting.plot_roi function to improve the image above.

````python
from nilearn import plotting

# 1. Clearer colors, higher visibility, and smooth interpolation
plotting.plot_roi(
    roi_img=GMprob_image,
    bg_img=T1_img,
    threshold=0.5,
    vmin=0.5,
    vmax=1.0,
    alpha=0.6,  # Increase from 0.2 so the overlay is clearly visible
    cmap="plasma",  # Smooth purple-to-yellow gradient (better than "jet")
    resampling_interpolation="continuous",  # Smooth edges instead of blocky pixels
    colorbar=True,
    title="1. Smooth Gradient & Higher Alpha (cmap='plasma', alpha=0.6)",
)

# 2. Contour outlines only (keeps the underlying T1 anatomy 100% visible)
plotting.plot_roi(
    roi_img=GMprob_image,
    bg_img=T1_img,
    threshold=0.5,
    view_type="contours",  # Draw boundary outlines instead of solid fill
    linewidths=0.8,
    cmap="autumn",
    title="2. Contour Outlines (view_type='contours')",
)

# 3. Multiple axial slices in a grid (mosaic view)
plotting.plot_roi(
    roi_img=GMprob_image,
    bg_img=T1_img,
    threshold=0.5,
    vmin=0.5,
    vmax=1.0,
    alpha=0.6,
    cmap="viridis",
    display_mode="mosaic",  # Show a grid of slices across the whole brain
    cut_coords=6,
    colorbar=True,
    title="3. Multi-Slice Mosaic View (display_mode='mosaic')",
)

# 4. Darker background contrast to make subtle probabilities pop
plotting.plot_roi(
    roi_img=GMprob_image,
    bg_img=T1_img,
    threshold=0.5,
    vmin=0.5,
    vmax=1.0,
    alpha=0.7,
    cmap="YlOrRd",
    dim=0.5,  # Dim the T1 background slightly for stronger contrast
    colorbar=True,
    title="4. Dimmed Background Contrast (dim=0.5, cmap='YlOrRd')",
)

plotting.show()
````
